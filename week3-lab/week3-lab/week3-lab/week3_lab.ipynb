### 2. What you are doing
Slide 25 walked one dial down a bowl, one step per click. Here is the same walk as a table. Then slide 27's four step sizes (Live demo 2), then your own, then slide 29's bumpy hill.

### 4. What you should see

**Table B**: 0.02 crawls (loss 1.59 after 40 steps), 0.1 and 0.9 converge, 1.98 oscillates, **2.05 diverges** (loss 396 and rising). No error was printed.
